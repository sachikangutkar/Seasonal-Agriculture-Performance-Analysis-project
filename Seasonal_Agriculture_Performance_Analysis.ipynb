## Problem Statement

## Objectives

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel("seasonal_agriculture_performance_dataset.xlsx")
print("Shape:", df.shape)
display(df.head())
display(df.info())
```

## 1. Data Quality Check

```python
print("Missing values:")
display(df.isna().sum()[df.isna().sum() > 0])

print("Duplicate rows:", df.duplicated().sum())
print("Season counts:")
display(df["Season"].value_counts())
```

### Cleaning
The supplied data contains missing values in `Rainfall_mm`, `Soil_Moisture_pct`, and `Yield_Tonnes_Ha`. Because these are numeric variables, missing values are replaced with their respective medians. Duplicate rows are not removed because the dataset contains no duplicates.


```python
clean = df.copy()
for col in ["Rainfall_mm", "Soil_Moisture_pct", "Yield_Tonnes_Ha"]:
    clean[col] = clean[col].fillna(clean[col].median())

clean["Profit_Margin_pct"] = np.where(
    clean["Revenue_INR"] != 0,
    clean["Profit_INR"] / clean["Revenue_INR"] * 100,
    np.nan
)
clean["Cost_per_Tonne_INR"] = np.where(
    clean["Production_Tonnes"] != 0,
    clean["Total_Cost_INR"] / clean["Production_Tonnes"],
    np.nan
)

print("Remaining missing values:", int(clean.isna().sum().sum()))
display(clean.head())
```

## 2. Seasonal Performance Analysis

```python
season_summary = clean.groupby("Season").agg(
    Farms=("Farm_ID","count"),
    Avg_Yield_Tonnes_Ha=("Yield_Tonnes_Ha","mean"),
    Total_Production_Tonnes=("Production_Tonnes","sum"),
    Avg_Revenue_INR=("Revenue_INR","mean"),
    Avg_Cost_INR=("Total_Cost_INR","mean"),
    Avg_Profit_INR=("Profit_INR","mean"),
    Avg_Water_Used_m3=("Water_Used_m3","mean"),
    Avg_Disease_Risk_pct=("Disease_Pest_Risk_pct","mean")
).round(2)

display(season_summary)
```

```python
plt.figure(figsize=(8,5))
plt.bar(season_summary.index, season_summary["Avg_Yield_Tonnes_Ha"])
plt.title("Average Yield by Season")
plt.xlabel("Season")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(season_summary.index, season_summary["Avg_Profit_INR"])
plt.title("Average Profit by Season")
plt.xlabel("Season")
plt.ylabel("Average Profit (INR)")
plt.tight_layout()
plt.show()
```

## 3. Crop Analysis

```python
crop_summary = clean.groupby("Crop").agg(
    Farms=("Farm_ID","count"),
    Avg_Yield_Tonnes_Ha=("Yield_Tonnes_Ha","mean"),
    Avg_Profit_INR=("Profit_INR","mean"),
    Avg_Disease_Risk_pct=("Disease_Pest_Risk_pct","mean")
).sort_values("Avg_Yield_Tonnes_Ha", ascending=False).round(2)

display(crop_summary)
```

```python
plt.figure(figsize=(9,5))
plt.bar(crop_summary.index, crop_summary["Avg_Yield_Tonnes_Ha"])
plt.title("Average Yield by Crop")
plt.xlabel("Crop")
plt.ylabel("Yield (Tonnes/Ha)")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()
```

## 4. Irrigation Analysis

```python
irrigation_summary = clean.groupby("Irrigation_Method").agg(
    Avg_Yield_Tonnes_Ha=("Yield_Tonnes_Ha","mean"),
    Avg_Profit_INR=("Profit_INR","mean"),
    Avg_Water_Used_m3=("Water_Used_m3","mean")
).sort_values("Avg_Yield_Tonnes_Ha", ascending=False).round(2)

display(irrigation_summary)
```

```python
plt.figure(figsize=(8,5))
plt.bar(irrigation_summary.index, irrigation_summary["Avg_Yield_Tonnes_Ha"])
plt.title("Average Yield by Irrigation Method")
plt.xlabel("Irrigation Method")
plt.ylabel("Yield (Tonnes/Ha)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

## 5. Environmental Relationship

```python
corr_cols = [
    "Rainfall_mm","Avg_Temperature_C","Humidity_pct","Sunlight_Hours_Day",
    "Soil_Moisture_pct","Nitrogen_kg_ha","Phosphorus_kg_ha","Potassium_kg_ha",
    "Fertilizer_kg_ha","Seed_Quality_Score","Yield_Tonnes_Ha",
    "Water_Used_m3","Disease_Pest_Risk_pct","Profit_INR"
]
corr = clean[corr_cols].corr(numeric_only=True)["Yield_Tonnes_Ha"].sort_values(ascending=False)
display(corr)
```

```python
plt.figure(figsize=(8,5))
plt.scatter(clean["Rainfall_mm"], clean["Yield_Tonnes_Ha"], alpha=0.35, s=12)
plt.title("Rainfall vs Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()
```

## 6. Key Findings

1. **Kharif has the highest average yield** among the three seasons in the supplied dataset.
2. **Kharif also has the highest average profit**, while Zaid has a negative average profit.
3. **Zaid uses the highest average amount of water**, while Rabi uses the lowest average water among the three seasons.
4. **Kharif shows the highest average disease/pest risk** in this dataset.
5. **Sugarcane has a much higher average yield** than the other crops, so crop comparisons should consider crop-specific production characteristics.
6. **Drip irrigation has the highest average yield** among the irrigation methods in the dataset.
7. The analysis identifies associations and patterns; it does **not** establish causal relationships.


## 7. Recommendations

- Use seasonal performance comparisons when planning crop and resource allocation.
- Investigate the higher disease/pest risk observed in Kharif and consider targeted monitoring.
- Evaluate drip irrigation practices for situations where water efficiency and yield are priorities.
- Review Zaid-season costs, water use and market outcomes because its average profit is negative.
- Analyze crop-specific economics rather than relying only on overall yield because crop types have very different yield scales.
- Continue collecting consistent environmental, resource and economic data to improve future seasonal planning.
